## Project Theme

Each row is one sensor reading with `temperature`, `vibration`, and `voltage`. The target `failure` marks whether the reading should trigger a maintenance warning.

The custom loader turns each row into a one-node graph. That sounds almost silly, but it is useful for learning: all graph-training machinery is exercised while the data shape stays easy to inspect by hand.

## What Lives Where

The examples live in `pioneerml_example_plugin.components_tutorial_examples`, grouped by pipeline phase:

- `data_loading`: input backend, graph loader, and loader manager.
- `modeling`: architecture, compiler, loss, Lightning module, and trainer.
- `evaluation`: evaluator, metric, and plot.
- `export`: exporter and model handle.
- `inference`: inference batch executor.
- `tuning`: HPO, objective, search space, and search parameter.
- `writing`: writer and output backend.
- `pipeline`: checked-in config plus helpers for loading or building config.
